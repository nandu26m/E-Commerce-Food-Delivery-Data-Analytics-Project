```python
# SECTION 1: Basics & Data Loading

import pandas as pd
print("Pandas version:", pd.__version__)
print("-" * 60)

# 1. Load orders.csv into a DataFrame
try:
    orders_df = pd.read_csv("orders.csv")
    print("Orders preview:\n", orders_df.head(1))
except Exception as e:
    print("Error loading orders.csv:", e)

print("-" * 60)

# 2. Load users.csv selecting only user_id and email
try:
    users_df = pd.read_csv("users.csv", usecols=["user_id", "email"])
    print("Users preview:\n", users_df.head(1))
except ValueError as e:
    print("Columns not found in users.csv:", e)
except FileNotFoundError:
    print("users.csv not found.")
except Exception as e:
    print("Error loading users.csv:", e)

print("-" * 60)

# 3. Load menu_items.csv into a DataFrame
try:
    menu_items_df = pd.read_csv("menu_items.csv")
    print("Menu items preview:\n", menu_items_df.head(1))
except Exception as e:
    print("Error loading menu_items.csv:", e)

print("-" * 60)

# 4. Load ratings.json into a DataFrame
try:
    ratings_df = pd.read_json("ratings.json")
    print("Ratings preview:\n", ratings_df.head(1))
except FileNotFoundError:
    print("ratings.json not found.")
except ValueError as e:
    print("Error parsing ratings.json:", e)
except Exception as e:
    print("Unexpected error loading ratings.json:", e)

print("-" * 60)

# 5. Display first 10 and last 10 rows of deliveries.csv
try:
    deliveries_df = pd.read_csv("deliveries.csv")
    print("First 10 deliveries:\n", deliveries_df.head(10))
    print("Last 10 deliveries:\n", deliveries_df.tail(10))
except Exception as e:
    print("Error loading deliveries.csv:", e)

print("-" * 60)

# 6. Show the shape of payments.csv
try:
    payments_df = pd.read_csv("payments.csv")
    print("Payments shape:", payments_df.shape)
except Exception as e:
    print("Error loading payments.csv:", e)

print("-" * 60)

# 7. Display column names and dtypes of restaurants.csv
try:
    restaurants_df = pd.read_csv("restaurants.csv")
    print("Restaurant dtypes:\n", restaurants_df.dtypes)
except Exception as e:
    print("Error loading restaurants.csv:", e)

print("-" * 60)

# 8. Check memory usage of orders.csv
try:
    print("Orders memory usage:\n", orders_df.memory_usage(deep=True))
except NameError:
    print("orders_df not loaded, skipping memory usage check.")
except Exception as e:
    print("Error checking memory usage:", e)

print("-" * 60)

# 9. Convert order_date column in orders.csv to datetime
try:
    if "order_date" in orders_df.columns:
        orders_df["order_date"] = pd.to_datetime(orders_df["order_date"], errors="coerce")
        print("order_date converted successfully.")
    else:
        print("Column 'order_date' not found in orders.csv.")
except Exception as e:
    print("Error converting order_date:", e)

print("-" * 60)

# 10. Convert price column in menu_items.csv to float
try:
    if "price" in menu_items_df.columns:
        menu_items_df["price"] = menu_items_df["price"].astype(float)
        print("Price column converted to float.")
    else:
        print("Column 'price' not found in menu_items.csv.")
except ValueError as e:
    print("Conversion error for 'price':", e)
except Exception as e:
    print("Error converting price column:", e)

print("-" * 60)

# 11. Load only the first 100 rows of promotions.csv
try:
    promotions_df = pd.read_csv("promotions.csv", nrows=100)
    print("Promotions shape (first 100 rows):", promotions_df.shape)
except FileNotFoundError:
    print("promotions.csv not found.")
except Exception as e:
    print("Error loading promotions.csv:", e)

print("-" * 60)
print("Section 1 completed successfully.")
```

```python
# SECTION 2: Data Cleaning
import pandas as pd
print(pd.__version__)

# Drop rows with missing email in users.csv.
import pandas as pd

try:
  users_df = pd.read_csv("users.csv")
  print("File loaded successfully. Total rows: ", len(users_df))

  if "email" in users_df.columns:
    initial_rows = len(users_df)
    users_df.dropna(subset=["email"], inplace=True)
    print(f"Dropped {initial_rows - len(users_df)} rows with missing email.")
  else:
    print("Column 'email' not found in the users.csv.")
except FileNotFoundError:
  print("File not found.")
except Exception as e:
  print("Error loading file 'users.csv': ", e)

# Fill missing feedback in ratings.csv with "No feedback".
import pandas as pd

try:
  # Load the data
  ratings_df = pd.read_csv("ratings.csv")
  print("File loaded. Total rows: ", len(ratings_df))

  # Check if "feedback" column exists.
  if "feedback" in ratings_df.columns:
    missing_before = ratings_df["feedback"].isna().sum()
    # Fill missing values
    ratings_df['feedback'] = ratings_df['feedback'].fillna("No feedback")
    missing_after = ratings_df["feedback"].isna().sum()
    print(f"Missing feedback before: {missing_before}")
    print(f"Missing feedback after: {missing_after}")
  else:
    print("column 'feedback' not found in ratings.csv")
except FileNotFoundError:
  print("File not found.")
except Exception as e:
  print("An error occured while processing ratings.csv", e)

# Replace negative total_amount in orders.csv with 0.
import pandas as pd

try:
    # Load the CSV
    orders_df = pd.read_csv("orders.csv")
    print("File loaded. Total rows:", len(orders_df))

    # Check if 'total_amount' column exists
    if 'total_amount' in orders_df.columns:
        negative_before = (orders_df['total_amount'] < 0).sum()
        # Replace negative values with 0
        orders_df['total_amount'] = orders_df['total_amount'].apply(lambda x: max(x, 0))
        negative_after = (orders_df['total_amount'] < 0).sum()
        print(f"Negative total_amount before: {negative_before}")
        print(f"Negative total_amount after: {negative_after}")
    else:
        print("Column 'total_amount' not found in orders.csv")

except FileNotFoundError:
    print("File 'orders.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'orders.csv' is empty.")
except Exception as e:
    print("An error occurred while processing orders.csv:", e)

# Remove duplicate rows in order_items.csv.
import pandas as pd

try:
    # Load the CSV
    order_items_df = pd.read_csv("order_items.csv")
    print("File loaded. Total rows:", len(order_items_df))

    # Remove duplicates
    duplicates_before = order_items_df.duplicated().sum()
    order_items_df = order_items_df.drop_duplicates()
    duplicates_after = order_items_df.duplicated().sum()

    print(f"Duplicate rows before: {duplicates_before}")
    print(f"Duplicate rows after: {duplicates_after}")

except FileNotFoundError:
    print("File 'order_items.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'order_items.csv' is empty.")
except Exception as e:
    print("An error occurred while processing order_items.csv:", e)

# Standardize restaurant_name in restaurants.csv (lowercase + strip).
import pandas as pd

try:
    # Load the CSV
    restaurants_df = pd.read_csv("restaurants.csv")
    print("File loaded. Total rows:", len(restaurants_df))

    # Check if 'restaurant_name' column exists
    if 'restaurant_name' in restaurants_df.columns:
        # Standardize: lowercase and strip spaces
        restaurants_df['restaurant_name'] = restaurants_df['restaurant_name'].str.lower().str.strip()
        print("restaurant_name standardized successfully.")
    else:
        print("Column 'restaurant_name' not found in restaurants.csv")

except FileNotFoundError:
    print("File 'restaurants.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'restaurants.csv' is empty.")
except Exception as e:
    print("An error occurred while processing restaurants.csv:", e)

# Rename columns first_name → fname and last_name → lname.
import pandas as pd

try:
    # Load the CSV
    users_df = pd.read_csv("users.csv")
    print("File loaded. Total rows:", len(users_df))

    # Check if the columns exist
    columns_to_rename = {}
    if 'first_name' in users_df.columns:
        columns_to_rename['first_name'] = 'fname'
    if 'last_name' in users_df.columns:
        columns_to_rename['last_name'] = 'lname'

    if columns_to_rename:
        users_df.rename(columns=columns_to_rename, inplace=True)
        print(f"Columns renamed: {columns_to_rename}")
    else:
        print("No columns to rename found in users.csv")

except FileNotFoundError:
    print("File 'users.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'users.csv' is empty.")
except Exception as e:
    print("An error occurred while processing users.csv:", e)

# Reorder columns in menu_items.csv as [menu_item_id, restaurant_id, name, price].
import pandas as pd

try:
    # Load the CSV
    menu_items_df = pd.read_csv("menu_items.csv")
    print("File loaded. Total rows:", len(menu_items_df))

    # Desired column order
    new_order = ['menu_item_id', 'restaurant_id', 'name', 'price']

    # Check which columns exist in the DataFrame
    existing_cols = [col for col in new_order if col in menu_items_df.columns]

    if existing_cols:
        menu_items_df = menu_items_df[existing_cols]
        print(f"Columns reordered to: {existing_cols}")
    else:
        print("None of the desired columns found in menu_items.csv")

except FileNotFoundError:
    print("File 'menu_items.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'menu_items.csv' is empty.")
except Exception as e:
    print("An error occurred while processing menu_items.csv:", e)

# Apply a custom function to clean promo_code in promotions.csv.
import pandas as pd

# Example custom function to clean promo codes
def clean_promo(code):
    if pd.isna(code):
        return "NO_CODE"
    return str(code).strip().upper()  # strip spaces and convert to uppercase

try:
    # Load the CSV
    promotions_df = pd.read_csv("promotions.csv")
    print("File loaded. Total rows:", len(promotions_df))

    # Check if 'promo_code' exists
    if 'promo_code' in promotions_df.columns:
        # Apply the custom function
        promotions_df['promo_code'] = promotions_df['promo_code'].apply(clean_promo)
        print("promo_code cleaned successfully.")
    else:
        print("Column 'promo_code' not found in promotions.csv")

except FileNotFoundError:
    print("File 'promotions.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'promotions.csv' is empty.")
except Exception as e:
    print("An error occurred while processing promotions.csv:", e)

# Replace infinite values in payments.csv with 0.
import pandas as pd
import numpy as np

try:
    # Load payments.csv
    payments_df = pd.read_csv("payments.csv")
    print("Payments loaded. Total rows:", len(payments_df))

    # Select numeric columns only
    numeric_cols = payments_df.select_dtypes(include=[np.number]).columns

    # Replace infinite values with 0
    payments_df[numeric_cols] = payments_df[numeric_cols].replace([np.inf, -np.inf], 0)

    print("Infinite values replaced with 0 in numeric columns.")

except FileNotFoundError:
    print("File 'payments.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'payments.csv' is empty.")
except Exception as e:
    print("An error occurred while processing payments.csv:", e)


# Normalize price column in menu_items.csv using Min-Max scaling.
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

try:
    # Load menu_items.csv
    menu_items_df = pd.read_csv("menu_items.csv")
    print("Menu items loaded. Total rows:", len(menu_items_df))

    # Check if 'price' column exists
    if "price" in menu_items_df.columns:
        scaler = MinMaxScaler()
        # Min-Max scaling requires a 2D array
        menu_items_df["price"] = scaler.fit_transform(menu_items_df[["price"]])
        print("Price column normalized using Min-Max scaling.")
    else:
        print("Column 'price' not found in menu_items.csv.")

except FileNotFoundError:
    print("File 'menu_items.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'menu_items.csv' is empty.")
except Exception as e:
    print("An error occurred while processing menu_items.csv:", e)

# Convert phone_number in users.csv to string type.
import pandas as pd

try:
    # Load users.csv
    users_df = pd.read_csv("users.csv")
    print("Users loaded. Total rows:", len(users_df))

    # Check if 'phone_number' column exists
    if "phone_number" in users_df.columns:
        users_df["phone_number"] = users_df["phone_number"].astype(str)
        print("phone_number converted to string type.")
    else:
        print("Column 'phone_number' not found in users.csv.")

except FileNotFoundError:
    print("File 'users.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'users.csv' is empty.")
except Exception as e:
    print("An error occurred while processing users.csv:", e)

# Fill missing delivery_time in deliveries.csv with the median.
import pandas as pd

try:
    # Load deliveries.csv
    deliveries_df = pd.read_csv("deliveries.csv")
    print("Deliveries loaded. Total rows:", len(deliveries_df))

    # Check if 'delivery_time' column exists
    if "delivery_time" in deliveries_df.columns:
        median_time = deliveries_df["delivery_time"].median()
        deliveries_df["delivery_time"] = deliveries_df["delivery_time"].fillna(median_time)
        print(f"Missing delivery_time filled with median value: {median_time}")
    else:
        print("Column 'delivery_time' not found in deliveries.csv.")

except FileNotFoundError:
    print("File 'deliveries.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'deliveries.csv' is empty.")
except Exception as e:
    print("An error occurred while processing deliveries.csv:", e)

# Drop columns not required for analysis in user_logins.csv.
import pandas as pd

try:
    # Load user_logins.csv
    user_logins_df = pd.read_csv("user_logins.csv")
    print("User logins loaded. Total rows:", len(user_logins_df))

    # Define columns to keep (only necessary for analysis)
    columns_to_keep = ["user_id", "login_timestamp"]  # modify as needed

    # Drop all other columns
    columns_to_drop = [col for col in user_logins_df.columns if col not in columns_to_keep]
    user_logins_df.drop(columns=columns_to_drop, inplace=True)

    print(f"Dropped columns not required for analysis: {columns_to_drop}")

except FileNotFoundError:
    print("File 'user_logins.csv' not found.")
except pd.errors.EmptyDataError:
    print("File 'user_logins.csv' is empty.")
except Exception as e:
    print("An error occurred while processing user_logins.csv:", e)
```
